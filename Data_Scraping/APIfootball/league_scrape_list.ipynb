```python
def league_obj(league_name, league_id, years):
    return {'name': league_name, 'id': league_id, 'years': years};

```

```python
import requests;

url = "https://v3.football.api-sports.io/leagues?country=";
api_key = "57199d6a3f17fa2348e9bb3eab7daf42";

payload={}
headers = {
  'x-rapidapi-key': api_key,
  'x-rapidapi-host': 'v3.football.api-sports.io'
}

dictionary = {};

for country in uefa_countries:
  dictionary[country] = [];
  response = requests.request("GET", url+f"{country}", headers=headers, data=payload)
  json_response = response.json()

  comp_num = len(json_response['response']);

  for i in range(0, comp_num):

    league_name = json_response['response'][i]['league']['name'];
    league_id = json_response['response'][i]['league']['id'];
    timespan = len(json_response['response'][i]['seasons'])
    years = [json_response['response'][i]['seasons'][j]['year'] for j in range(timespan)]

    invalid_input = True;
    while(invalid_input):
      ans = input(f"Include {league_name} from {country}? (y/n) ")

      if ans == 'y':
        dictionary[country].append(league_obj(league_name, league_id, years))
        invalid_input=False
      elif ans == "n":
        invalid_input =False
        continue
      else:
        print("invalid response.")
```
